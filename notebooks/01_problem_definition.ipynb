# Electrical Grid Stability

## Problem Context

This project focuses on analyzing the stability of a 4-node electrical grid system.

The main problem is to understand whether the electrical system remains stable under different operating conditions.

Understanding the stability of the system is important for assessing its operating condition and identifying potentially unstable conditions.

## Intended Users

The results could be useful for power system engineers and companies involved in operating and managing electrical grids.

## Expected Output

The output should indicate whether the electrical grid is stable or unstable based on the system's operating conditions and features.
